```python

import cv2
import matplotlib.pyplot as plt
```

```python
path ="test.jpg"
image = cv2.imread(path)
```

```python
image_rgb= cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
```

```python
plt.imshow(image_rgb)
plt.axis('off')#hides the x
plt.show()
```

```python
gray= cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
plt.imshow(gray)
plt.axis('off')#hides the x
plt.show()
```

```python
blurred_image = cv2.GaussianBlur(image_rgb, (15, 15), 0)
plt.imshow(blurred_image)
plt.axis('off')#hides the x
plt.show()
```

```python
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
edges = cv2.Canny(gray, 100, 200)
edges_rgb = cv2.cvtColor(edges, cv2.COLOR_GRAY2RGB)
plt.imshow(edges)
plt.axis('off')#hides the x
plt.show()
```

```python
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
edges = cv2.Canny(gray, 100, 200)
edges_rgb = cv2.cvtColor(edges, cv2.COLOR_GRAY2RGB)
plt.imshow(edges_rgb)
plt.axis('off')#hides the x
plt.show()
```

```python
bright = cv2.convertScaleAbs(image_rgb, beta=60)
plt.imshow(bright)
plt.axis('off')#hides the x
plt.show()
```

```python
contrast = cv2.convertScaleAbs(image_rgb, alpha=1.1)
plt.imshow(contrast)
plt.axis('off')
plt.show()
```

```python
rotated = cv2.rotate(image_rgb, cv2.ROTATE_90_CLOCKWISE)
plt.imshow(rotated)
plt.axis('off')
plt.show()
```

```python
flipped = cv2.flip(image_rgb, 0)
plt.imshow(flipped)
plt.axis('off')
plt.show()
```
